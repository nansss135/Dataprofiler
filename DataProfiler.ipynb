```python
#Distribution of Age
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=20, kde=True, color="pink")
plt.title("Distribution of Age")
plt.show()
```

```python
#Distribution of Income
plt.figure(figsize=(8,5))
sns.histplot(df["Income"], bins=20, kde=True, color="pink")
plt.title("Distribution of Income")
plt.show()
```

```python
#Distribution of Purchases
plt.figure(figsize=(8,5))
sns.histplot(df["Purchases"], bins=20, kde=True, color="pink")
plt.title("Distribution of Purchases")
plt.show()
```

```python
#Gender vs Purchases
plt.figure(figsize=(8,5))
sns.boxplot(x="Gender", y="Purchases", data=df)
plt.title("Gender vs Purchases")
plt.show()
```

```python
#Income vs Churn
plt.figure(figsize=(8,5))
sns.boxplot(x="Churn", y="Income", data=df)
plt.title("Income vs Churn")
plt.show()
```

```python
#Membership Status vs Purchases
plt.figure(figsize=(8,5))
sns.boxplot(x="Membership_Status", y="Purchases", data=df)
plt.title("Membership Status vs Purchases")
plt.show()
```

```python
#Churn Count
plt.figure(figsize=(6,4))
sns.countplot(x="Churn", data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
#Correlation Heatmap
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Pair Plot
sns.pairplot(
    df[["Age","Income","Purchases","Purchase_Frequency"]]
)
plt.show()
```
